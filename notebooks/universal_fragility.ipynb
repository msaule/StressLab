# StressLab Universal Fragility Study

This notebook sketches the discovery-engine workflow for generating synthetic systems, running collapse campaigns, and producing theory and research artifacts.


```python
from pathlib import Path

from stresslab.generator import SyntheticGenerationConfig, build_generation_artifacts
from stresslab.discovery import build_discovery_artifacts
from stresslab.research import build_research_artifacts, build_theory_artifacts

study_root = Path('runs/notebook_universal_fragility')
study_root.mkdir(parents=True, exist_ok=True)
```

```python
generation = build_generation_artifacts(
    study_root / 'generate',
    config=SyntheticGenerationConfig(count=25, topology_type='mixed', seed=17),
)
generation
```

```python
discover = build_discovery_artifacts(study_root / 'discover', generation_config=SyntheticGenerationConfig(count=25, topology_type='mixed', seed=17))
discover
```

```python
theory = build_theory_artifacts(study_root / 'discover', output_dir=study_root / 'theory')
research = build_research_artifacts(study_root / 'discover', output_dir=study_root / 'research')
theory, research
```
